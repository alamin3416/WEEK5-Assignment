```python
import numpy as np
import pandas as pd

# 1. Data Generation (Seed 11)
np.random.seed(11)
n = 288 # 24 hours, one reading every 5 minutes
times = pd.date_range("2026-10-13 00:00", periods=n, freq="5min")
hours = times.hour + times.minute / 60
truth_temp = 20 + 2.5 * np.sin((hours - 9) / 24 * 2 * np.pi) + np.random.normal(0, 0.1, n)
truth = pd.DataFrame({"timestamp": times, "temp_c": truth_temp})

sensor = truth.copy()
drop = np.random.choice(sensor.index, size=34, replace=False) # ~12% lost packets
sensor = sensor.drop(index=drop).reset_index(drop=True)
stuck_start = 150 # sensor freezes for 14 readings
sensor.loc[stuck_start:stuck_start + 13, "temp_c"] = sensor.loc[stuck_start, "temp_c"]
safe = [i for i in sensor.index if not (stuck_start <= i <= stuck_start + 13)]
g = np.random.choice(safe, size=3, replace=False) # 3 glitches
sensor.loc[g, "temp_c"] = [-40.0, 85.0, 0.0]

print(f"Initial arrived readings: {len(sensor)}")

# ==========================================
# EDGE STAGE (Cheap checks only: local range checks)
# ==========================================
# The edge device filters out physically impossible outlier values locally
# to save bandwidth and prevent transmission of corrupt data.
edge_cleaned = sensor.copy()
valid_range_mask = (edge_cleaned["temp_c"] >= 5.0) & (edge_cleaned["temp_c"] <= 40.0)
edge_cleaned.loc[~valid_range_mask, "temp_c"] = np.nan

# ==========================================
# CLOUD STAGE (Heavier steps: history-based checks, stuck detection, imputation)
# ==========================================
# The cloud reindexes to the full 5-minute time grid, detects stuck sensors
# using rolling standard deviation, and applies interpolation.
cloud_df = edge_cleaned.set_index("timestamp").reindex(times).reset_index()

# Detect stuck sensor: rolling standard deviation of 0 over a window of 6 readings
rolling_std = cloud_df["temp_c"].rolling(window=6, min_periods=6).std()
stuck_mask = rolling_std == 0.0

# Mark stuck values as NaN so they can be interpolated
cloud_df.loc[stuck_mask, "temp_c"] = np.nan

# Interpolate missing values and stuck blocks using linear interpolation
cloud_df["temp_c"] = cloud_df["temp_c"].interpolate(method="linear")

print(f"Cloud cleaned readings: {len(cloud_df)}")
```
